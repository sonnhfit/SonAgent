```python
from agno.team import Team
from agno.agent import Agent
from agno.models.openai import OpenAIResponses
from agno.tools.hackernews import HackerNewsTools
from agno.tools.yfinance import YFinanceTools

from agno.agent import Agent, RunOutput

from agno.db.sqlite import SqliteDb
from agno.models.anthropic import Claude
from agno.os import AgentOS
from agno.tools.mcp import MCPTools
from agno.models.openai import OpenAIResponses
from agno.utils.pprint import pprint_run_response
from agno.models.groq import Groq



model = Groq(id="openai/gpt-oss-20b", api_key="")


news_agent = Agent(
    name="News Agent",
    model=model,
    role="Get trending tech news from HackerNews",
    tools=[HackerNewsTools()]
)

finance_agent = Agent(
    name="Finance Agent",
    model=model,
    role="Get stock prices and financial data",
    tools=[YFinanceTools()]
)

team = Team(
    name="Research Team",
    members=[news_agent, finance_agent],
    model=model,
    instructions="Delegate to the appropriate agent based on the request.",
    debug_mode=True
)

team.print_response("What are the trending AI stories and how is NVDA stock doing?", stream=False)
```

```python
response: RunOutput = team.run("Trending startups and products.")
```
